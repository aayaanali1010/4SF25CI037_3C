```python
#1. Importing libraries and reading the dataset
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("imdb_clean.csv")

df.head()
```

```python
#3. Data transformation
#A. Dropping unnecessary column
df = df.drop('Unnamed: 0', axis=1)

print(df.head())
```

```python
#B. Changing data types
df['release_year'] = df['release_year'].astype(int)

df['runtime'] = df['runtime'].astype(int)

df['metascore'] = df['metascore'].astype(int)

df.head()
```

```python
#C. Adding a new column – Runtime in hours
df['runtime_hours'] = df['runtime'] / 60

print(df[['title', 'runtime', 'runtime_hours']].head())
```

```python
#D. Adding a new column – Release decade
df['decade'] = (df['release_year'] // 10) * 10

print(df[['title', 'release_year', 'decade']].head())
```

```python
#4. Handling null values
#A. Checking null values
df.isnull().sum()


```

```python
#B. Filling null values using mean
df['rating'] = df['rating'].fillna(df['rating'].mean())

df['metascore'] = df['metascore'].fillna(df['metascore'].mean())

df['gross(M)'] = df['gross(M)'].fillna(df['gross(M)'].mean())

df['runtime'] = df['runtime'].fillna(df['runtime'].mean())
```

```python
#C. Checking null values after filling
df.isnull().sum()
```

```python
#D. Using apply and lambda
df['double_rating'] = df['rating'].apply(lambda x: x * 2)

print(df[['title', 'rating', 'double_rating']].head())
```

```python
#5. Insight generation
movies = df.drop_duplicates(subset=['title', 'director', 'release_year'])
```

```python
#A. Average rating by genre
genre_rating = df.groupby('genre')['rating'].mean()

print(genre_rating)
```

```python
#B. Top 10 genres by number of movies
genre_count = df.groupby('genre')['title'].nunique()

genre_count = genre_count.sort_values(ascending=False)

print(genre_count.head(10))
```

```python
#C. Number of movies released in each decade
decade_movies = movies.groupby('decade')['title'].count()

print(decade_movies)
```

```python
#D. Top 10 highest-rated movies
top_rated = movies.sort_values('rating', ascending=False)

print(top_rated[['title', 'rating', 'director']].head(10))
```

```python
#E. Top 10 highest-grossing movies
top_gross = movies.sort_values('gross(M)', ascending=False)

print(top_gross[['title', 'gross(M)', 'rating']].head(10))

```

```python
#F. Average rating by director
director_rating = movies.groupby('director')['rating'].mean()

director_rating = director_rating.sort_values(ascending=False)

print(director_rating.head(10))
```

```python
#G. Average runtime and rating by decade
decade_analysis = movies.groupby('decade')[['runtime', 'rating']].mean()

print(decade_analysis)
```

```python
#6. Data visualization
#A. Top 10 genres – Bar chart
top_genres = genre_count.head(10)

plt.figure(figsize=(10,5))

top_genres.plot(kind='bar')

plt.title("Top 10 Movie Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Movies")

plt.show()
```

```python
#B. Movies released per decade – Line graph
plt.figure(figsize=(10,5))

decade_movies.plot(kind='line', marker='o')

plt.title("Movies Released Per Decade")
plt.xlabel("Decade")
plt.ylabel("Number of Movies")

plt.grid()
plt.show()
```

```python
#C. Movie rating distribution – Histogram
plt.figure(figsize=(8,5))

plt.hist(movies['rating'], bins=10)

plt.title("Distribution of IMDb Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Movies")

plt.show()
```

```python
#D. Rating vs Gross Revenue – Scatter plot
plt.figure(figsize=(8,5))

plt.scatter(movies['rating'], movies['gross(M)'])

plt.title("Movie Rating vs Gross Revenue")
plt.xlabel("IMDb Rating")
plt.ylabel("Gross Revenue (Millions)")

plt.show()
```

```python
#E. Average rating by genre – Bar chart
plt.figure(figsize=(12,5))

genre_rating.sort_values(ascending=False).plot(kind='bar')

plt.title("Average IMDb Rating by Genre")
plt.xlabel("Genre")
plt.ylabel("Average Rating")

plt.show()
```

```python
#7. Save the transformed dataset
df.to_csv("imdb_transformed.csv", index=False)

print("Dataset saved successfully!")
```
